# Course progress

Tick what you have worked through, write notes as you go, and draw the board.

Your ticks and notes are saved to `progress.json` the moment you make them, so
you can close this notebook and come back to it. The section list is read from
the module articles each time rather than stored, so it cannot drift out of
step with the course.

Run the cells in order.

````python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd()))

from tools import progress

modules = progress.discover()
print(f"{len(modules)} modules, "
      f"{sum(len(m.items) for m in modules)} sections and exercises")
````

## 1. Tick and take notes

One tab per module. Each tab holds:

- a **status** dropdown, left on `auto` unless you want to override it
- a **tick box** per section, plus one for the exercise
- a **notes** box for that module

The last tab holds the notes that appear on the board itself.

On `auto`, the status is derived: nothing ticked is *Not started*, everything
ticked is *Completed*, anything between is *Partially covered*. Ticks cannot
tell *In progress* from *Partially covered*, since both are some-but-not-all,
so that is the one call left to you. Set it in the dropdown.

````python
progress.checklist(modules);
````

## 2. Draw the board

Re-run this cell any time to pick up what you have just ticked.

````python
progress.render()
````

## 3. Keep a copy

Writes a standalone HTML file you can open in a browser or share. It carries its
own dark background, so it looks the same wherever it is opened.

````python
import pathlib

out = pathlib.Path("progress.html")
out.write_text(progress.build_html(), encoding="utf-8")
print(f"wrote {out.resolve()}")
````
